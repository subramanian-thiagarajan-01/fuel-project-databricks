```sql
CREATE OR REFRESH STREAMING LIVE TABLE fuel_project_dev.gold.fact_google_reviews AS
SELECT 
  review_id,
  customer_id,
  fuel_station_id,
  rating,
  review,
  CASE 
    WHEN rating >= 4 THEN 'Positive'
    WHEN rating = 3 THEN 'Neutral'
    ELSE 'Negative'
  END as sentiment,
  CASE WHEN review IS NULL THEN 1 ELSE 0 END as is_review_missing,
  length(review) as review_length,
  date_format(modified_ts, 'yyyyMMdd') as date_key,
  modified_ts,
  current_timestamp() as processed_at
FROM STREAM(fuel_project_dev.silver.google_reviews);
```
