```sql
CREATE MATERIALIZED VIEW fuel_project_dev.gold.mv_station_daily_performance AS
SELECT 
  station_id,
  date_key,
  COUNT(DISTINCT fill_id) as total_transactions,
  SUM(fuel_volume) as total_volume_liters,
  SUM(revenue) as total_revenue,
  AVG(revenue) as avg_revenue_per_transaction,
  AVG(fuel_volume) as avg_volume_per_transaction,
  COUNT(DISTINCT CASE WHEN fuel_type = 'Petrol' THEN fill_id END) as petrol_count,
  COUNT(DISTINCT CASE WHEN fuel_type = 'Diesel' THEN fill_id END) as diesel_count,
  COUNT(DISTINCT CASE WHEN fuel_type = 'CNG' THEN fill_id END) as cng_count,
  SUM(CASE WHEN fuel_type = 'Petrol' THEN revenue ELSE 0 END) as petrol_revenue,
  SUM(CASE WHEN fuel_type = 'Diesel' THEN revenue ELSE 0 END) as diesel_revenue,
  SUM(CASE WHEN fuel_type = 'CNG' THEN revenue ELSE 0 END) as cng_revenue
FROM fuel_project_dev.gold.fact_fuel_transactions
GROUP BY station_id, date_key;
```
