```sql
CREATE OR REPLACE TABLE fuel_project_dev.gold.dim_time AS
SELECT 
  hour,
  CASE 
    WHEN hour BETWEEN 6 AND 11 THEN 'Morning'
    WHEN hour BETWEEN 12 AND 17 THEN 'Afternoon'
    WHEN hour BETWEEN 18 AND 21 THEN 'Evening'
    ELSE 'Night'
  END as time_of_day,
  CASE 
    WHEN hour BETWEEN 7 AND 9 OR hour BETWEEN 17 AND 19 THEN 'Peak'
    ELSE 'Off-Peak'
  END as peak_category
FROM (SELECT explode(sequence(0, 23)) as hour);
```
