```sql
CREATE OR REPLACE VIEW fuel_project_dev.gold.vw_batch_station_performance AS
SELECT 
  d.fuel_station_id,
  d.manager_name,
  d.station_size_category,
  d.employee_count,
  p.date_key,
  dt.full_date,
  dt.day_name,
  dt.day_type,
  p.total_transactions,
  p.total_revenue,
  p.total_volume_liters,
  p.avg_revenue_per_transaction,
  p.petrol_revenue,
  p.diesel_revenue,
  p.cng_revenue,
  r.avg_rating,
  r.positive_reviews,
  r.negative_reviews,
  ROUND(p.total_revenue / d.employee_count, 2) as revenue_per_employee
FROM fuel_project_dev.gold.mv_station_daily_performance p
INNER JOIN fuel_project_dev.gold.dim_fuel_station d ON p.station_id = d.fuel_station_id
INNER JOIN fuel_project_dev.gold.dim_date dt ON p.date_key = dt.date_key
LEFT JOIN fuel_project_dev.gold.mv_station_review_metrics r 
  ON p.station_id = r.fuel_station_id 
  AND p.date_key = r.date_key;
```
