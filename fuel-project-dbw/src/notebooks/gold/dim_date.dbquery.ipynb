```sql
CREATE OR REPLACE TABLE fuel_project_dev.gold.dim_date AS
SELECT 
  date_format(date, 'yyyyMMdd') as date_key,
  date as full_date,
  year(date) as year,
  quarter(date) as quarter,
  month(date) as month,
  weekofyear(date) as week_of_year,
  dayofweek(date) as day_of_week,
  dayofmonth(date) as day_of_month,
  date_format(date, 'EEEE') as day_name,
  date_format(date, 'MMMM') as month_name,
  CASE WHEN dayofweek(date) IN (1,7) THEN 'Weekend' ELSE 'Weekday' END as day_type
FROM (
  SELECT explode(sequence(to_date('2024-01-01'), to_date('2026-12-31'), interval 1 day)) as date
);
```
