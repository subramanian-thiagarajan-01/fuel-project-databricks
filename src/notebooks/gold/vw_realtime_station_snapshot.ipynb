```python
%sql
CREATE OR REPLACE VIEW fuel_project_dev.gold.vw_realtime_station_snapshot AS
SELECT 
  d.fuel_station_id,
  d.manager_name,
  d.station_size_category,
  d.latitude,
  d.longitude,
  COUNT(DISTINCT f.fill_id) as transactions_last_hour,
  SUM(f.revenue) as revenue_last_hour,
  SUM(f.fuel_volume) as volume_last_hour,
  AVG(f.transaction_duration_sec) as avg_transaction_time_sec,
  MAX(f.start_time) as last_transaction_time
FROM fuel_project_dev.gold.dim_fuel_station d
LEFT JOIN fuel_project_dev.gold.fact_fuel_transactions f
  ON d.fuel_station_id = f.station_id
  AND f.start_time >= current_timestamp() - interval 1 hour
GROUP BY d.fuel_station_id, d.manager_name, d.station_size_category, d.latitude, d.longitude;
```
