```python
%sql
CREATE OR REPLACE TABLE fuel_project_dev.gold.dim_fuel_station AS
SELECT 
  fuel_station_id,
  latitude,
  longitude,
  manager_name,
  fuel_support,
  employee_count,
  area_sqft,
  CASE 
    WHEN area_sqft < 6000 THEN 'Small'
    WHEN area_sqft < 8000 THEN 'Medium'
    ELSE 'Large'
  END as station_size_category,
  current_timestamp() as last_updated
FROM fuel_project_dev.silver.fuel_station;
```
