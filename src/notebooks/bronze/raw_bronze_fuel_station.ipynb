```python
from pyspark.sql.functions import *
```

```python
SOURCE_VOLUME = "/Volumes/fuel_project_dev/raw/fuel_station_metadata/"
TARGET_TABLE = "fuel_project_dev.bronze.fuel_station"
```

```python
df = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")   # ✅ schema can evolve
        .load(SOURCE_VOLUME)
)
```

```python
display(df)
```

```python
df = (
    df
        .withColumnRenamed("Fuel Station ID", "fuel_station_id")
        .withColumnRenamed("Location", "location")
        .withColumnRenamed("Manager Name", "manager_name")
        .withColumnRenamed("Fuel Support", "fuel_support")
        .withColumnRenamed("EmployeeCount", "employee_count")
        .withColumnRenamed("Area Sqft", "area_sqft")
        .withColumn("_ingestion_ts", current_timestamp())
        .withColumn("_source_file", col("_metadata.file_path"))
)
```

```python
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable(TARGET_TABLE)
```
